```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 100)

y = np.sin(x)

sine_data = pd.DataFrame({
    "X": x,
    "Sine_Value": y
})

sine_data.to_csv(
    "sine_wave_data.csv",
    index=False
)

plt.plot(
    sine_data["X"],
    sine_data["Sine_Value"]
)

plt.title("Sine Wave")
plt.xlabel("X")
plt.ylabel("sin(X)")
plt.grid(True)

plt.show()
```
