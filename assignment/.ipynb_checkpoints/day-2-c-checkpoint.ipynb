```python
import pandas as pd
import re

df = pd.read_csv("student_data.csv")

print("Original Dataset:")
print(df)

# Remove extra spaces
df["Name"] = df["Name"].str.strip()
df["City"] = df["City"].str.strip()
df["Email"] = df["Email"].str.strip()

# Remove special characters from names
df["Name"] = df["Name"].str.replace(
    r"[^A-Za-z ]", "", regex=True
)

# Standardize names
df["Name"] = df["Name"].str.title()

# Standardize city names
df["City"] = df["City"].str.title()

# Convert email to lowercase
df["Email"] = df["Email"].str.lower()

# Validate email using RegEx
df["Valid_Email"] = df["Email"].str.match(
    r"^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$"
)

# Convert Age to numeric
df["Age"] = pd.to_numeric(
    df["Age"], errors="coerce"
)

# Replace unrealistic ages
df.loc[(df["Age"] < 15) | (df["Age"] > 100), "Age"] = None

# Convert Marks to numeric
df["Marks"] = pd.to_numeric(
    df["Marks"], errors="coerce"
)

# Replace invalid marks
df.loc[(df["Marks"] < 0) | (df["Marks"] > 100), "Marks"] = None

# Fill missing numerical values
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

# Remove invalid email records
df = df[df["Valid_Email"] == True]

# Remove duplicate records
df = df.drop_duplicates()

# Remove helper column
df = df.drop(columns=["Valid_Email"])

# Reset index
df = df.reset_index(drop=True)

print("\nCleaned Dataset:")
print(df)

df.to_csv("cleaned_student_data.csv", index=False)
```
