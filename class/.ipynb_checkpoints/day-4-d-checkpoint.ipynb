```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from wordcloud import WordCloud
import matplotlib.pyplot as plt
from collections import Counter

nltk.download('punkt')
nltk.download('stopwords')

text = """
One of my favorite hobbies is reading manga and fanfiction novels. I enjoy reading different kinds of stories because they allow me to explore new worlds, characters, and ideas. Manga is especially interesting to me because it combines storytelling with visual art, making the characters and events easier to imagine. I also enjoy fanfiction novels because they take familiar characters or fictional worlds and present them in new situations and stories. Sometimes, fanfiction explores alternative endings, different character relationships, or completely new storylines that are not present in the original work.Reading these stories is a great way for me to relax during my free time. It also helps improve my imagination and creativity because I often think about how the story might continue or how I would change certain events. Manga and fanfiction introduce me to different writing styles, cultures, and perspectives. They also help me develop my vocabulary and reading skills naturally. Another reason I enjoy this hobby is that there are many genres to choose from, including action, fantasy, science fiction, comedy, mystery, and adventure. Overall, reading manga and fanfiction is both an enjoyable form of entertainment and a way for me to explore my creativity and imagination.
"""

print(text)

words = word_tokenize(text.lower())
print(words)

stop_words = set(stopwords.words('english'))

filtered_words = [
    word for word in words
    if word.isalpha() and word not in stop_words
]

print(filtered_words)

word_freq = Counter(filtered_words)
print(word_freq)

wordcloud = WordCloud(
    width=800,
    height=400,
    background_color='white'
).generate_from_frequencies(word_freq)

plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.show()
```
