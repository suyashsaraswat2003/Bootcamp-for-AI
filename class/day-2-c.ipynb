```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    'Singer': ['Arijit Singh', 'Shreya Ghoshal', 'Sonu Nigam', 'Lata Mangeshkar', 'Kishore Kumar'],
    'Years_Active': [15, 20, 30, 70, 40],
    'Number_of_Songs': [500, 700, 600, 1000, 800],
    'Awards_Won': [50, 60, 40, 100, 80]
}

singers_df = pd.DataFrame(data)
print(singers_df)
```

```python
singers_df.info()
print(singers_df.tail(2))
```

```python
print(singers_df.describe())

print("Maximum songs:", singers_df['Number_of_Songs'].max())
print("Minimum songs:", singers_df['Number_of_Songs'].min())
print("Average awards:", singers_df['Awards_Won'].mean())
print("Median years active:", singers_df['Years_Active'].median())
```

```python
print(singers_df.sort_values(by='Awards_Won', ascending=False))

print(singers_df[singers_df['Awards_Won'] > 50])
```

```python
plt.figure(figsize=(8, 5))
plt.bar(singers_df['Singer'], singers_df['Number_of_Songs'])
plt.title('Number of Songs by Indian Singers')
plt.xlabel('Singer')
plt.ylabel('Number of Songs')
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.show()
```
