```python
import numpy as np

# Synthetic 2D Dataset
data = np.array([
    [45, 50, 55, 60, 65],
    [52, 58, 63, 68, 72],
    [67, 70, 75, 78, 82],
    [71, 74, 79, 83, 88],
    [83, 85, 89, 91, 95]
])

print("Dataset:")
print(data)

# Shape and Dimensions
print("\nShape:", data.shape)
print("Number of Dimensions:", data.ndim)
print("Total Elements:", data.size)

# Sum
print("\nSum of All Elements:", np.sum(data))
print("Sum Along Axis 0:", np.sum(data, axis=0))
print("Sum Along Axis 1:", np.sum(data, axis=1))

# Mean
print("\nMean of All Elements:", np.mean(data))
print("Mean Along Axis 0:", np.mean(data, axis=0))
print("Mean Along Axis 1:", np.mean(data, axis=1))

# Median
print("\nMedian of All Elements:", np.median(data))
print("Median Along Axis 0:", np.median(data, axis=0))
print("Median Along Axis 1:", np.median(data, axis=1))

# Minimum
print("\nMinimum of All Elements:", np.min(data))
print("Minimum Along Axis 0:", np.min(data, axis=0))
print("Minimum Along Axis 1:", np.min(data, axis=1))

# Maximum
print("\nMaximum of All Elements:", np.max(data))
print("Maximum Along Axis 0:", np.max(data, axis=0))
print("Maximum Along Axis 1:", np.max(data, axis=1))

# Range
print("\nRange of All Elements:", np.ptp(data))
print("Range Along Axis 0:", np.ptp(data, axis=0))
print("Range Along Axis 1:", np.ptp(data, axis=1))

# Variance
print("\nVariance of All Elements:", np.var(data))
print("Variance Along Axis 0:", np.var(data, axis=0))
print("Variance Along Axis 1:", np.var(data, axis=1))

# Standard Deviation
print("\nStandard Deviation of All Elements:", np.std(data))
print("Standard Deviation Along Axis 0:", np.std(data, axis=0))
print("Standard Deviation Along Axis 1:", np.std(data, axis=1))

# Percentiles
print("\n25th Percentile:", np.percentile(data, 25))
print("50th Percentile:", np.percentile(data, 50))
print("75th Percentile:", np.percentile(data, 75))

print("\n25th Percentile Along Axis 0:", np.percentile(data, 25, axis=0))
print("50th Percentile Along Axis 0:", np.percentile(data, 50, axis=0))
print("75th Percentile Along Axis 0:", np.percentile(data, 75, axis=0))

print("\n25th Percentile Along Axis 1:", np.percentile(data, 25, axis=1))
print("50th Percentile Along Axis 1:", np.percentile(data, 50, axis=1))
print("75th Percentile Along Axis 1:", np.percentile(data, 75, axis=1))

# Quartiles
Q1 = np.percentile(data, 25, axis=0)
Q2 = np.percentile(data, 50, axis=0)
Q3 = np.percentile(data, 75, axis=0)

print("\nQ1 Along Axis 0:", Q1)
print("Q2 Along Axis 0:", Q2)
print("Q3 Along Axis 0:", Q3)

Q1 = np.percentile(data, 25, axis=1)
Q2 = np.percentile(data, 50, axis=1)
Q3 = np.percentile(data, 75, axis=1)

print("\nQ1 Along Axis 1:", Q1)
print("Q2 Along Axis 1:", Q2)
print("Q3 Along Axis 1:", Q3)

# Interquartile Range
print("\nIQR Along Axis 0:", 
      np.percentile(data, 75, axis=0) - np.percentile(data, 25, axis=0))

print("IQR Along Axis 1:", 
      np.percentile(data, 75, axis=1) - np.percentile(data, 25, axis=1))

# Cumulative Sum
print("\nCumulative Sum Along Axis 0:")
print(np.cumsum(data, axis=0))

print("\nCumulative Sum Along Axis 1:")
print(np.cumsum(data, axis=1))

# Cumulative Product
print("\nCumulative Product Along Axis 0:")
print(np.cumprod(data, axis=0))

print("\nCumulative Product Along Axis 1:")
print(np.cumprod(data, axis=1))

# Difference
print("\nDifference Along Axis 0:")
print(np.diff(data, axis=0))

print("\nDifference Along Axis 1:")
print(np.diff(data, axis=1))

# Absolute Deviation from Mean
mean_axis0 = np.mean(data, axis=0)
mean_axis1 = np.mean(data, axis=1)

print("\nAbsolute Deviation from Mean Along Axis 0:")
print(np.abs(data - mean_axis0))

print("\nAbsolute Deviation from Mean Along Axis 1:")
print(np.abs(data - mean_axis1[:, np.newaxis]))

# Mean Absolute Deviation
print("\nMean Absolute Deviation Along Axis 0:")
print(np.mean(np.abs(data - mean_axis0), axis=0))

print("\nMean Absolute Deviation Along Axis 1:")
print(np.mean(np.abs(data - mean_axis1[:, np.newaxis]), axis=1))

# Z-Score Along Axis 0
z_axis0 = (data - np.mean(data, axis=0)) / np.std(data, axis=0)

print("\nZ-Score Along Axis 0:")
print(z_axis0)

# Z-Score Along Axis 1
z_axis1 = (data - np.mean(data, axis=1, keepdims=True)) / np.std(data, axis=1, keepdims=True)

print("\nZ-Score Along Axis 1:")
print(z_axis1)

# Sorting
print("\nSorted Along Axis 0:")
print(np.sort(data, axis=0))

print("\nSorted Along Axis 1:")
print(np.sort(data, axis=1))

# Index of Minimum
print("\nIndex of Minimum Along Axis 0:")
print(np.argmin(data, axis=0))

print("\nIndex of Minimum Along Axis 1:")
print(np.argmin(data, axis=1))

# Index of Maximum
print("\nIndex of Maximum Along Axis 0:")
print(np.argmax(data, axis=0))

print("\nIndex of Maximum Along Axis 1:")
print(np.argmax(data, axis=1))

# Covariance
print("\nCovariance Matrix:")
print(np.cov(data))

print("\nCovariance Matrix of Columns:")
print(np.cov(data, rowvar=False))

# Correlation
print("\nCorrelation Matrix:")
print(np.corrcoef(data))

print("\nCorrelation Matrix of Columns:")
print(np.corrcoef(data, rowvar=False))
```
